```python
# Only keep users who rated at least X books
min_ratings_user = 50
active_users = ratings_with_books['User-ID'].value_counts()
active_users = active_users[active_users >= min_ratings_user].index

# Only keep books that have at least X ratings
min_ratings_book = 50
popular_books = ratings_with_books['Book-Title'].value_counts()
popular_books = popular_books[popular_books >= min_ratings_book].index

# Filter the dataframe
filtered = ratings_with_books[
    (ratings_with_books['User-ID'].isin(active_users)) &
    (ratings_with_books['Book-Title'].isin(popular_books))
]

print("Users:", filtered['User-ID'].nunique())
print("Books:", filtered['Book-Title'].nunique())

# Now pivot — much smaller
book_pivot = filtered.pivot_table(
    columns='User-ID',
    index='Book-Title',
    values='Book-Rating'
)
```

```python
book_pivot
```

```python
from scipy.sparse import csr_matrix

book_sparse = csr_matrix(book_pivot.fillna(0))
```

```python
book_sparse
```

```python
from sklearn.neighbors import NearestNeighbors
model = NearestNeighbors(metric='cosine', algorithm='brute')
model.fit(book_sparse)
```

```python
distance,suggestion = model.kneighbors(book_pivot.iloc[237].values.reshape(1,-1),n_neighbors=6)
```

```python
url_list = []
for name in book_name[0]:
    ids = np.where(ratings_with_books['Book-Title'] == name)[0][0]
    url = ratings_with_books.iloc[ids]['Image-URL-M']
    url_list.append(url)
    
```
